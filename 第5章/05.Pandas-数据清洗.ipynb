```python
import pandas as pd

pd.set_option('display.min_rows', 30)  # 设置显示行数(不设置默认10行)

# 1. 读取数据
df = pd.read_csv('data/sales.csv')

# 2. 数据清洗
# 2.1 缺失值处理
# df.isnull()  # 查看缺失值

# 2.1.1 删除缺失值
# df.dropna()  # 删除缺失值所在行
# df = df.dropna(axis=1)  # 删除缺失值所在列
# df

# 2.1.2 填充缺失值
df.fillna('--')  # 填充缺失值为'--'
df.ffill()  # 前向填充, 用前一行的值填充缺失值
df.bfill()  # 后向填充, 用后一行的值填充缺失值
```

```python
# 2.2 重复值处理
# 2.2.1 查看重复值
df.duplicated()  # 查看重复值所在行(默认检查所有列)
df.duplicated(subset=['订单号'])  # 查看重复值所在行(只检查指定列)

# 2.2.2 删除重复值
df.drop_duplicates(subset=['订单号'], keep='last') # keep='first' 保留第一个重复值, keep='last' 保留最后一个重复值, keep=False 删除所有重复值
```

```python
# 2.3 异常值处理
# 2.3.1 查看异常值
df[df['单价'] < 0]

# 2.3.2 删除异常值
df.drop(df[df['单价'] < 0].index)  # 删除异常值所在行

# 2.3.3 修复异常值
df['单价'] = df['单价'].abs() # 将单价的值转换为绝对值
df
```

```python
# 2.4 数据格式处理
df['订单日期'] = df['订单日期'].str.replace('/', '-')
df
```
